# Assignment 7: Understanding Loss Functions

### Objective
Understand how loss functions measure the difference between expected outputs and model predictions using Mean Squared Error and Binary Cross-Entropy.

## Task 1: Define Actual and Predicted Values (Prepare the Data)

A small binary classification example is used to compare actual class labels with predicted probabilities.

```python
import numpy as np

# Actual class labels
true_labels = np.array([1, 0, 1, 0, 1])

# Probabilities predicted by the model
model_output = np.array([0.85, 0.25, 0.70, 0.15, 0.90])

print("Actual Labels:", true_labels)
print("Predicted Values:", model_output)
```

# Task 2: Calculate Mean Squared Error

### (MSE)
MSE calculates the average squared difference between the actual and predicted values.

### The formula is:

MSE = (1/n) × Σ(Actual − Predicted)²

```python
# Calculate Mean Squared Error
mse_value = np.mean((true_labels - model_output) ** 2)

print("Mean Squared Error:", round(mse_value, 4))
```

# Task 3: Binary Cross-Entropy
Binary Cross-Entropy evaluates how closely the predicted probabilities match the actual binary classes.


For binary classification, the formula is:

Cross-Entropy Loss = −(1/n) × Σ[y log(p) + (1−y) log(1−p)]

```python
tiny = 1e-15
safe_output = np.clip(model_output, tiny, 1 - tiny)

bce_value = -np.mean(
    true_labels * np.log(safe_output) +
    (1 - true_labels) * np.log(1 - safe_output)
)

print("Binary Cross-Entropy:", round(bce_value, 4))
```

# Task 4: Compare the Results

MSE measures squared numerical differences, whereas Binary Cross-Entropy focuses on the probabilities assigned to the correct class.

```python
print("Loss Comparison")
print("----------------")
print(f"MSE                  : {mse_value:.4f}")
print(f"Binary Cross-Entropy : {bce_value:.4f}")
```

# Comparison

| Feature | Mean Squared Error (MSE) | Binary Cross-Entropy (BCE) |
|---|---|---|
| Main purpose | Measures squared difference between actual and predicted values | Measures how well predicted probabilities match actual classes |
| Commonly used for | Regression problems | Binary classification problems |
| Output type | Numerical value | Probability-based classification loss |
| Error calculation | Squares the difference between values | Uses logarithmic penalty for incorrect probabilities |
| Sensitivity | Larger errors are penalized more strongly | Strongly penalizes confident incorrect predictions |
| Example application | Predicting house prices or temperature | Spam detection or yes/no classification |
| In this experiment | Measures the difference between labels and predictions | Measures the quality of the predicted class probabilities |

### Result Comparison

The calculated MSE and Binary Cross-Entropy values are different because they use different methods to measure prediction error. MSE focuses on the numerical difference between actual and predicted values, while Binary Cross-Entropy evaluates the predicted probabilities for the correct class.



## When Can Each Loss Function Be Used?

- **MSE:** commonly used for regression problems involving numerical predictions.
- **Binary Cross-Entropy:** commonly used for binary classification where the model produces probabilities for two classes.



# Task 5: Visualization

```python
import matplotlib.pyplot as plt

loss_names = ["MSE", "Binary Cross-Entropy"]
loss_values = [mse_value, bce_value]

plt.figure(figsize=(7, 5))
plt.bar(loss_names, loss_values)
plt.title("Comparison of Loss Functions")
plt.xlabel("Loss Function")
plt.ylabel("Loss Value")
plt.tight_layout()
plt.show()
```
